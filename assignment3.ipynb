```python
#to check if given triangle is right angle triangle while using function

def checker(a,b,c) : 
    sum_squarea = a*a + b*b
    sum_squareb = c**2 + b*b
    sum_squarec = a*a + c**2
    if sum_squarea  == c**2 :
        print("right angle triangle!")
    elif sum_squareb == a**2 :
        print("right angle triangle!")
    elif sum_squarec == b**2 :
        print("right angle triangle!")
    else:
        print("Not an right angle triangle!")


num_1 = int(input("a:"))
num_2 = int(input("b:"))
num_3 = int(input("c"))
checker (num_1,num_2,num_3)
```
